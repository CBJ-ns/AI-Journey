```python
!pip install ultralytics
from ultralytics import YOLO
model = YOLO('yolov8s-seg.pt')
results = model('https://ultralytics.com/images/bus.jpg')
results[0].save('result.jpg')
results2 = model('测试.jpg',conf=0.1)
results2[0].save('my_result_force.jpg')
results3 = model('qunxiang.png')
results3[0].save('qunxiang_result.png')
print('模型在这张图里看到的候选目标：',results[0].boxes.cls)
#  把置信度门槛降到0.01，逼它画出所有可疑的框
results_low_conf = model('测试.jpg',conf=0.01)
results_low_conf[0].save('my_result_001.jpg')
```

```python
class MyCrackModel:
  def __init__(self,model_name):
    self.model_name = model_name
    print(f'系统已加载模型：{self.model_name}')
  def detect(self,image_name):
    print(f'模型{self.model_name}正在处理图片：{image_name}')
    return {"crack":True,"confidence":0.99}
crack_detector = MyCrackModel("YOLOv8-Crack")
result = crack_detector.detect('图片.jpg')
print(f'输出结果：{result}')
```
